```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Nayal\Downloads\Projet Kaggle\archive\supermarket_sales.csv")
df
```

```python
# Ventes totales par moyen de paiement
df.groupby('payment_method')['revenue'].sum().plot(kind='bar', color='steelblue')
plt.title('Ventes totales par moyen de paiement')
plt.xlabel('Moyen de paiement')
plt.ylabel('Total')
plt.tight_layout()
plt.show()
```

```python
# Note moyenne des clients par ville
df.groupby('city')['rating'].mean().plot(kind='bar', color='coral')
plt.title('Note moyenne des clients par ville')
plt.xlabel('Ville')
plt.ylabel('Note moyenne')
plt.tight_layout()
plt.show()
```

```python
# Chiffre d'affaires par catégorie de produit
df.groupby('product_line')['revenue'].sum().sort_values().plot(kind='barh', color='green')
plt.title("Chiffre d'affaires par catégorie")
plt.xlabel('Revenu total')
plt.tight_layout()
plt.show()
```

```python
# Statistiques de base
print("=== RÉSUMÉ DES VENTES ===")
print(f"Chiffre d'affaires total : {df['revenue'].sum().round(2)}")
print(f"Vente moyenne par transaction : {df['revenue'].mean().round(2)}")
print(f"Vente maximale : {df['revenue'].max().round(2)}")
print(f"Vente minimale : {df['revenue'].min().round(2)}")
print(f"Nombre de transactions : {len(df)}")
```

```python
# Évolution des ventes par date
df['date'] = pd.to_datetime(df['date'])
df.groupby('date')['revenue'].sum().plot(figsize=(12,4), color='purple')
plt.title('Évolution des ventes dans le temps')
plt.xlabel('Date')
plt.ylabel('Revenu')
plt.tight_layout()
plt.show()
```

```python
# Répartition des ventes par genre
df.groupby('gender_customer')['revenue'].sum().plot(kind='pie', autopct='%1.1f%%', colors=['lightblue', 'pink'])
plt.title('Répartition des ventes par genre')
plt.ylabel('')
plt.tight_layout()
plt.show()
```
